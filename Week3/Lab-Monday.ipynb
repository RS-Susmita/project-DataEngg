```python
# Q12:  Create data/source/customers.csv with the supplied data.

import os
import shutil
import csv

# Create folders
os.makedirs("data/source", exist_ok=True)
os.makedirs("data/raw", exist_ok=True)

# Create source file with supplied data
data = """customer_id,name,email,city
1, Amelia Clarke ,AMELIA@EXAMPLE.COM,Manchester
2,Oliver Bennett,oliver@example.com,Bristol
"""

with open("data/source/customers.csv", "w", newline="") as f:
    f.write(data)

# Copy unchanged to data/raw
shutil.copyfile("data/source/customers.csv", "data/raw/customers.csv")

# Count copied data rows (excluding header)
with open("data/raw/customers.csv", newline="") as f:
    reader = csv.reader(f)
    row_count = sum(1 for _ in reader) - 1

print(f"Number of copied data rows: {row_count}")
```

```python
# Q13: Clean Raw Customers into Staging

# Create staging folder
os.makedirs("data/staging", exist_ok=True)

cleaned_rows = []

with open("data/raw/customers.csv", newline="") as f:
    reader = csv.DictReader(f)
    fieldnames = reader.fieldnames
    for row in reader:
        row["name"] = row["name"].strip()
        row["email"] = row["email"].strip().lower()
        cleaned_rows.append(row)

with open("data/staging/customers.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(cleaned_rows)

# Print cleaned rows
for row in cleaned_rows:
    print(row)
```

```python
#Q. 20 : Stop a Pipeline after Failure

#Create a Python runner with these functions:
#ingest() returns success.
#clean() raises FileNotFoundError when data/raw/customers.csv is missing.
#publish() runs only when earlier functions succeed.
#The runner prints the failed task name and stops.

from pathlib import Path

def ingest():
    print("Ingestion has done successfully")

def clean():
    raw_file = Path("data/raw/customers.csv")
    if not raw_file.exists():
        raise FileNotFoundError(raw_file)
    print("cleaned")

def publish():
    print("publish:success")


def run_pipeline():
    tasks = [
        ("ingest", ingest),
        ("clean", clean),
        ("publish", publish),
    ]

    for task_name, task_func in tasks:
        try:
            task_func()
        except Exception as e:
            print(f"Task failed: {task_name} ({e})")
            print("Pipeline stopped.")
            return  # stop the pipeline here

    print("Pipeline completed successfully.")


if __name__ == "__main__":
    run_pipeline()
```
