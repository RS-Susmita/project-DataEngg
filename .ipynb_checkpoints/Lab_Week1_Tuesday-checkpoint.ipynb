```python
# Lab 15

product_ids = [701, 702, 703]
order_ids = [801, 802, 801, 803]
transactions = [10, -5, 10, 20]

immutable_products = tuple(product_ids)
removed_duplicates = set(order_ids)
transactions_list = list(transactions)

print("Immutable product IDs:", immutable_products)
print("Unique order IDs for quick membership:", removed_duplicates)
print("Transactions list (order preserved, duplicates allowed):", transactions_list)
```

```python
# Lab 16

orders = [
    {"order_id": 1, "customer": "Asha", "amount": 250.0, "product_id": 1001, "status": "delivered"},
    {"order_id": 2, "customer": "Ravi", "amount": 100.0, "product_id": 1002, "status": "pending"},
    {"order_id": 3, "customer": "Asha", "amount": 150.0, "product_id": 1003, "status": "delivered"},
    {"order_id": 4, "customer": "Meera", "amount": 400.0, "product_id": 1001, "status": "delivered"},
    {"order_id": 5, "customer": "Ravi", "amount": 50.0, "product_id": 1002, "status": "cancelled"}
]


totals_by_customer = {}
unique_products = set()
delivered_count = 0
 
for order in orders:
    if order["status"]!= "delivered":
        continue
    delivered_count+= 1
    Customer = order["customer"]
    mount = order["amount"]
    product_id = order["product_id"]
 
    # aggregate totals by customer
    if customer in totals_by_customer:
        totals_by_customer[customer] += amount
    else:
        totals_by_customer[customer] = amount
 
    # add product to set
    unique_products.add(product_id)
 
# Prepare report dictionary
report = {
    "totals_by_customer": totals_by_customer,
    "unique_products": unique_products,
    "delivered_count": delivered_count
}
 
# Print report in readable form
print("Totals by customer:")
for cust, amt in report["totals_by_customer"].items():
    print(f"{cust}: {amt}")
 
print("\nUnique products sold (delivered):", report["unique_products"])
print("Delivered order count:", report["delivered_count"])

```
