```python
# Q3. Load coffee_sales.csv into a DataFrame and find the total revenue.

df = (spark.read
      .format("csv")
      .option("header", "true")
      .option("inferSchema", "true")
      .load("/Volumes/workspace/default/data_assignment2/coffee_sales.csv"))

display(df.limit(10))
total_revenue = df.agg({"money": "sum"})
display(total_revenue)
```

```python
# Q4. Find the number of transactions for each coffee_name using groupBy.

coffee_counts = df.groupBy("coffee_name").count()

# Display the results
display(coffee_counts)
```

```python
# Q5. Add a new column for month from the date column, and find total revenue per month.

from pyspark.sql.functions import month, to_date, sum

df_with_month = df.withColumn("month", month(to_date("date", "yyyy-MM-dd")))
#display(df_with_month.limit(10))
# 2. Group by the new 'month' column and sum up 'money'
monthly_revenue = df_with_month.groupBy("month").agg(sum("money").alias("total_revenue"))

# Display the results
display(monthly_revenue.sort("month"))
```
