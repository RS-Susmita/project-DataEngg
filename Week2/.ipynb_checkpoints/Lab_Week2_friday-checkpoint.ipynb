```python
import sqlite3
import random

DB_NAME = "query_performance.db"

# ---------------------------------------------------------------
# 1. Connect (creates query_performance.db if it doesn't exist)
# ---------------------------------------------------------------
conn = sqlite3.connect(DB_NAME)
cur = conn.cursor()

# ---------------------------------------------------------------
# 2. Create the orders table
# ---------------------------------------------------------------
cur.execute("DROP TABLE IF EXISTS orders")
cur.execute("""
   CREATE TABLE orders (
       order_id    INTEGER PRIMARY KEY,
       customer_id INTEGER NOT NULL,
       amount_gbp  REAL NOT NULL,
       status      TEXT NOT NULL
   )
""")

# ---------------------------------------------------------------
# 3. Insert 5,000 sample orders
# ---------------------------------------------------------------
statuses = ["pending", "shipped", "delivered", "cancelled", "refunded"]
num_customers = 500  # customers repeat across orders, so indexing helps

sample_orders = [
   (
       random.randint(1, num_customers),               # customer_id
       round(random.uniform(5.0, 500.0), 2),            # amount_gbp
       random.choice(statuses),                         # status
   )
   for _ in range(5000)
]

cur.executemany(
   "INSERT INTO orders (customer_id, amount_gbp, status) VALUES (?, ?, ?)",
   sample_orders,
)
conn.commit()

# ---------------------------------------------------------------
# 4. EXPLAIN QUERY PLAN BEFORE creating the index
# ---------------------------------------------------------------
test_query = "SELECT * FROM orders WHERE customer_id = 250"

cur.execute(f"EXPLAIN QUERY PLAN {test_query}")
plan_before = cur.fetchall()

# ---------------------------------------------------------------
# 5. Create an index on customer_id
# ---------------------------------------------------------------
cur.execute("CREATE INDEX IF NOT EXISTS idx_orders_customer_id ON orders (customer_id)")
conn.commit()

# ---------------------------------------------------------------
# 6. EXPLAIN QUERY PLAN AFTER creating the index
# ---------------------------------------------------------------
cur.execute(f"EXPLAIN QUERY PLAN {test_query}")
plan_after = cur.fetchall()

# ---------------------------------------------------------------
# 7. Print both query plans
# ---------------------------------------------------------------
print(f"Query used for EXPLAIN QUERY PLAN:\n  {test_query}\n")

print("=" * 60)
print("QUERY PLAN AFTER INDEX ON customer_id")
print("=" * 60)
for row in plan_after:
   print(row)

conn.close()
```

```python
# Question 4: 

import sqlite3
import random
import time

DB_NAME = "query_performance.db"

conn = sqlite3.connect(DB_NAME)
cur = conn.cursor()

cur.execute("DROP TABLE IF EXISTS orders")
cur.execute("""
    CREATE TABLE orders (
        order_id    INTEGER PRIMARY KEY,
        customer_id INTEGER NOT NULL,
        amount_gbp  REAL NOT NULL,
        status      TEXT NOT NULL
    )
""")

statuses = ["Pending", "Dispatched", "Delivered", "Cancelled", "Refunded"]
num_customers = 500

sample_orders = [
    (
        random.randint(1, num_customers),
        round(random.uniform(5.0, 500.0), 2),
        random.choice(statuses),
    )
    for _ in range(5000)
]

cur.executemany(
    "INSERT INTO orders (customer_id, amount_gbp, status) VALUES (?, ?, ?)",
    sample_orders,
)
conn.commit()

# Approach A: Read every order, filter + sum in Python
start_a = time.perf_counter()
cur.execute("SELECT status, amount_gbp FROM orders")
all_rows = cur.fetchall()

total_a = 0.0
for status, amount in all_rows:
    if status == "Dispatched":
        total_a += amount
time_a = time.perf_counter() - start_a

# Approach B: Filter + sum in SQL, only the final value returned
start_b = time.perf_counter()
cur.execute("SELECT SUM(amount_gbp) FROM orders WHERE status = 'Dispatched'")
total_b = cur.fetchone()[0]
time_b = time.perf_counter() - start_b

# Results
print("APPROACH A — filter & sum in Python")
print(f"Rows fetched from DB : {len(all_rows)}")
print(f"Total (Dispatched)   : £{total_a:,.2f}")
print(f"Time taken           : {time_a:.6f} sec")

print("\nAPPROACH B — filter & sum in SQL (SUM)")
print(f"Rows fetched from DB : 1 (just the total)")
print(f"Total (Dispatched)   : £{total_b:,.2f}")
print(f"Time taken           : {time_b:.6f} sec")

if round(total_a, 2) == round(total_b, 2):
    print(f"\n✅ MATCH: Both approaches agree — total = £{total_a:,.2f}")
else:
    print(f"\n❌ MISMATCH: A=£{total_a:,.2f}  B=£{total_b:,.2f}")

conn.close()
```
