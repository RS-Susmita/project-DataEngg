```python
# Lab 8

def apply_tax(amounts, tax_rate=0.1):
    new_list = []

    for amount in amounts:
        tax = amount*tax_rate
        total_amount = amount + tax
        new_list.append(total_amount)
    return new_list

amounts = [100,200,350]

print(apply_tax(amounts))
print(apply_tax(amounts,0.05))

    
    
```

```python
# Lab 9

def process_score(score, max_score=100):
    if score < 0 or score  > max_score:
        return None
        
    percent = (score/max_score)*100
    return round(percent,1)

print(process_score(85))
print(process_score(150))
print(process_score(-5,50))
print(process_score(5,50))
    
    
```

```python
# Lab 15

raw_sales = [120.0, 250.5, np.nan, -20.0, 400.0, 150.0, 0.0, 300.0]

def clean_sales(arr):
    # Converts input list to a NumPy array of floats
    Float_array = np.array(arr,dtype = float)
    Replaced_array = Float_array.copy()
    
    #Replaces negative and zero values with np.nan
    Replaced_array[Replaced_array <= 0] = np.nan

    #Removes np.nan values and returns the cleaned array
    Cleaned_array =  Replaced_array[~np.isnan(Replaced_array)]
    
    return Float_array , Replaced_array, Cleaned_array
   


def apply_discount_to_large_sales(arr, threshold=200, discount_rate=0.1):

# Apply discount to values greater than threshold
    discounted = arr.copy()
    discounted[discounted > threshold] = discounted[discounted > threshold] * (1 - discount_rate)

    return discounted


def summarize_sales(arr):
    total = arr.sum()
    count = len(arr)
    mean = arr.mean()
    minimum = arr.min()
    maximum = arr.max()

    return {
        "Total": total,
        "Count": count,
        "Mean": mean,
        "Minimum": minimum,
        "Maximum": maximum
    }

Float_Array, Replaced_Array, Cleaned_Array = (clean_sales(raw_sales))  
Discounted_Array = apply_discount_to_large_sales(Cleaned_Array)   
Summary = summarize_sales(Discounted_Array)




print("Float_Array = ",Float_Array)
print("Replaced_Array =", Replaced_Array)
print ("Cleaned Array = ", Cleaned_Array)
print ("Discounted Array = ", Discounted_Array)
print(f"Total: {Summary['Total']:.2f}")
print(f"Count: {Summary['Count']}")
print(f"Mean: {Summary['Mean']:.2f}")
print(f"Min: {Summary['Minimum']:.1f}")
print(f"Max: {Summary['Maximum']:.1f}")




```
