```python
# Data Understanding
#1. Load the dataset into a DataFrame called df. 
#2. Display the first 5 rows. 
#3. Check the number of rows and columns. 
#4. List all column names. 

import pandas as pd
# 1. Load data set
df = pd.read_csv("~/Susmita/Projects/Data_Engg/Class_Works/Data/nba.csv")
# 2. Display first five rows
print( df.head(5))
# 3. check number of rows and column
print("Shape= ", df.shape)
# 4. List all column names
print ("Column Names = ", df.columns)
```

```python
# Data Cleaning 
#  5. Convert the Salary column to numeric (handle errors). 
Salary_numerics = df["Salary"] = pd.to_numeric(df["Salary"])#, errors="coerce")
#print(Salary_numerics)

#  6. Count how many missing (NaN) values are in each column. 
#print ("Count_nan = \n column "" number\n", df.isnull().sum())

#  7. Drop rows where Salary is missing. 
df.dropna(subset=["Salary"])
#print(miss_salary.head(5))

#  8. Fill missing College values with "Unknown". 
df["College"] = df["College"].fillna("Unknown")
print(df.head(5))
```

```python
#Filtering Data
#9. Show all players from Boston Celtics. 

Boston = df[df["Team"]== "Boston Celtics"]
#print(Boston)

#10. Show players whose Age is greater than 30. 

AgeOver30 = df[df["Age"]>30]
#print(AgeOver30)

#11. Filter players who play as PG (Point Guard). 

PG_Players = df[df["Position"]=="PG"]
#print(PG_Players)

#12. Find players with Salary greater than 5,000,000. 
High_salary_players = df[df["Salary"]>5000000]
print(High_salary_players)

print(len(Boston))
print(len(AgeOver30))
print(len(PG_Players))
print(len(High_salary_players))
```

```python
 #sorting
 # 13. Sort players by Salary (highest first). 
sort_salary = df.sort_values(by="Salary",ascending=False)
#print(sort_salary.head(10))

 # 14. Sort players by Age (youngest first). 
#sort_age = df.sort_values(by="Age",ascending=True)
sort_age = sort_salary.sort_values(by="Age",ascending=True)
print(sort_age.head(10))
```

```python
# GroupBy Operations 
#15. Find the average salary per team. 

average_salary = df.groupby("Team")["Salary"].mean()
#print (average_salary.head(5))
average_salary_college = df.groupby("College")["Salary"].mean()
#print(average_salary_college.head(5))

#    16. Count number of players in each team. 
Number_Player_Team = df.groupby("Team")["Name"].count()
#print("Count =", Number_Player_Team.head(5))

#    17. Find the maximum salary in each team. 
Max_salary_team = df.groupby("Team")["Salary"].max()
Max_Salary =print (" Team " "     "   "Max Salary \n", Max_salary_team.head(5))
```

```python
# Aggregations
#   18. What is the average age of players? 

Avg_age = df["Age"].mean()
print("Average Age =", Avg_age)

# 19. What is the maximum salary in the dataset? 
Max_salary = df["Salary"].max()
print("Max_Salary = ", Max_salary)

# 20. What is the minimum weight? 
Min_Wt = df["Weight"].min()
print ("Minimum_Wt = ", Min_Wt)

```

```python
# Column Operations
#21. Create a new column Age_in_5_years. 
df["Age_in_5_years"] = df["Age"] +5
#print(df.head(5))

#    22. Create a new column Salary_in_Millions (Salary / 100000). 

df["Salary_in_Millions"] = df["Salary"]/100000
print(df.head(5))
```
